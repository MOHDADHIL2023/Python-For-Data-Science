#1. Simple Sales Transaction


```python
import pandas as pd

data = {
    'Product':['Widget A', 'Widget B', 'Widget C', 'Widget D', 'Widget E'],
    'Date':['2026-01-01', '2026-01-03', '2026-01-03', '2026-01-04', '2026-01-05'],
    'Sales':[150, 200, 175, 300, 250]
}

df =pd.DataFrame(data)
df
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
import matplotlib.pyplot as plt

plt.plot(df['Date'], df['Sales'], marker='o')
plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
df.groupby('Product')['Sales'].sum().plot(kind='bar')
plt.title('Total Sales by Product')
plt.show()
```

```python
df.groupby('Product')['Sales'].sum().plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales Distribution')
plt.ylabel('')
plt.show()
```

```python
import plotly.express as px

fig = px.bar(df, x='Product', y='Sales', title='Sales by Product')
fig.show()
```

#2. Employees Performance Metrics

```python
import pandas as pd

data = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Years_Experience': [2, 5, 8, 1, 3],
    'Performance_Score': [85, 92, 78, 90, 88]
}

df = pd.DataFrame(data)
df
```

```python
print(df[['Years_Experience', 'Performance_Score']].corr())
```

```python
print(df.loc[df['Performance_Score'].idxmax()])
```

```python
print(df.loc[df['Years_Experience'].idxmax()])
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['Years_Experience'], df['Performance_Score'], color='blue')

for i in range(len(df)):
    plt.text(df['Years_Experience'][i], df['Performance_Score'][i], df['Employee'][i])

plt.title('Experience vs Performance')
plt.xlabel('Years of Experience')
plt.ylabel('Performance Score')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(x='Years_Experience', y='Performance_Score', data=df)
plt.title('Experience vs Performance (Regression)')
plt.show()
```

```python
df.plot(x='Employee', y='Performance_Score', kind='bar', color='green')
```

```python
import matplotlib.pyplot as plt

sizes = df['Performance_Score'] * 5  # bubble size

plt.scatter(df['Years_Experience'], df['Performance_Score'], s=sizes, alpha=0.5)

plt.title('Employee Performance Bubble Chart')
plt.xlabel('Years of Experience')
plt.ylabel('Performance Score')
plt.show()
```

#3. Regional Marketing Data

```python
import pandas as pd

data = {
    'Region': ['North', 'South', 'East', 'West'],
    'Budget': [5000, 4500, 7000, 6000],
    'Actual_Spend': [4800, 4900, 6500, 5800]
}

df = pd.DataFrame(data)
df
```

```python
df['Variance'] = df['Budget'] - df['Actual_Spend']
df
```

```python
df['Spend_Ratio'] = (df['Actual_Spend'] / df['Budget']) * 100
df
```

```python
df.loc[df['Actual_Spend'].idxmax()]
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(len(df['Region']))
width = 0.35

plt.bar(x - width/2, df['Budget'], width, label='Budget')
plt.bar(x + width/2, df['Actual_Spend'], width, label='Actual Spend')

plt.xticks(x, df['Region'])
plt.title('Budget vs Actual Spend by Region')
plt.ylabel('Amount')
plt.legend()
plt.show()
```

```python
df.set_index('Region')['Budget'].plot(kind='pie', autopct='%1.1f%%')
plt.title('Marketing Budget Distribution')
plt.ylabel('')
plt.show()
```

```python
df.set_index('Region')['Actual_Spend'].plot(kind='pie', autopct='%1.1f%%')
plt.title('Actual Spend Distribution')
plt.ylabel('')
plt.show()
```

```python
df['Variance'].plot(kind='bar', color='orange')
plt.title('Budget Variance by Region')
plt.xlabel('Region')
plt.ylabel('Budget - Actual Spend')
plt.show()
```

#4. Monthly Website Traffic

```python
import pandas as pd

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Visitors': [1200, 1500, 1400, 1800, 2100],
    'Conversion_Rate': [0.02, 0.025, 0.022, 0.03, 0.035]
}

df = pd.DataFrame(data)
df
```

```python
df['Visitor_Growth'] = df['Visitors'].diff()
df
```

```python
df['Conversion_Rate'].mean()
```

```python
df.loc[df['Visitors'].idxmax()]
```

```python
import matplotlib.pyplot as plt

plt.plot(df['Month'], df['Visitors'], marker='o', color='blue')
plt.title('Monthly Website Visitors')
plt.xlabel('Month')
plt.ylabel('Visitors')
plt.grid(True)
plt.show()
```

```python
plt.fill_between(df['Month'], df['Visitors'], color='skyblue', alpha=0.4)
plt.plot(df['Month'], df['Visitors'], color='blue')
plt.title('Monthly Website Traffic (Area Chart)')
plt.xlabel('Month')
plt.ylabel('Visitors')
plt.show()
```

```python
fig, ax1 = plt.subplots()

ax1.plot(df['Month'], df['Visitors'], color='blue', marker='o')
ax1.set_xlabel('Month')
ax1.set_ylabel('Visitors', color='blue')

ax2 = ax1.twinx()
ax2.plot(df['Month'], df['Conversion_Rate'], color='red', marker='o')
ax2.set_ylabel('Conversion Rate', color='red')

plt.title('Visitors vs Conversion Rate')
plt.show()
```

```python
plt.plot(df['Month'], df['Conversion_Rate'], marker='o', color='green')
plt.title('Monthly Conversion Rate Trend')
plt.xlabel('Month')
plt.ylabel('Conversion Rate')
plt.grid(True)
plt.show()
```

```python
df['Growth_%'] = df['Visitors'].pct_change() * 100
df
```

#5. Inventory Stock Levels

```python
import pandas as pd

data = {
    'Item': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Stock_Level': [50, 200, 150, 80],
    'Reorder_Point': [20, 50, 50, 30]
}

df = pd.DataFrame(data)
df
```

```python
df[df['Stock_Level'] < df['Reorder_Point']]
```

```python
df['Excess_Stock'] = df['Stock_Level'] - df['Reorder_Point']
df
```

```python
df['Stock_%'] = (df['Stock_Level'] / (df['Reorder_Point'] * 2)) * 100
df
```

```python
import matplotlib.pyplot as plt

plt.bar(df['Item'], df['Stock_Level'], color='skyblue')
plt.title('Inventory Stock Levels')
plt.xlabel('Item')
plt.ylabel('Stock Level')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(len(df['Item']))
width = 0.35

plt.bar(x - width/2, df['Stock_Level'], width, label='Stock Level')
plt.bar(x + width/2, df['Reorder_Point'], width, label='Reorder Point')

plt.xticks(x, df['Item'])
plt.title('Stock vs Reorder Point')
plt.legend()
plt.show()
```

```python
colors = ['red' if df['Stock_Level'][i] < df['Reorder_Point'][i] else 'green'
          for i in range(len(df))]

plt.bar(df['Item'], df['Stock_Level'], color=colors)
plt.title('Stock Levels (Red = Needs Reorder)')
plt.xlabel('Item')
plt.ylabel('Stock Level')
plt.show()
```

```python
for i in range(len(df)):
    print(f"{df['Item'][i]} → {df['Stock_Level'][i]}/{df['Reorder_Point'][i]}")
```

```python
import plotly.graph_objects as go

fig = go.Figure(go.Indicator(
    mode = "gauge+number",
    value = df['Stock_Level'][0],
    gauge={'axis': {'range': [0, 200]}},
    title={'text': df['Item'][0]}
))
fig.show()
```
